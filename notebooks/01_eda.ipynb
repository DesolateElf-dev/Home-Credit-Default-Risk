# Creating database postgre

```python
import pandas as pd
from sqlalchemy import create_engine
from sqlalchemy.engine import URL
import glob

url = URL.create(
    drivername="postgresql+psycopg2",
    username="postgres",
    password="eduardo@adm",
    host="localhost",
    port=5432,
    database="Home-Credit-Default-Risk"
)

engine = create_engine(url)

# Teste a conexão antes de importar
with engine.connect() as conn:
    print("Conexão OK!")

arquivos = glob.glob('C:/Users/emaci/VSCode/Home-Credit-Default-Risk/data/home-credit-default-risk/application_test.csv/*.csv')
print(f"{len(arquivos)} arquivo(s) encontrado(s)")

for arquivo in arquivos:
    df = pd.read_csv(arquivo)
    nome_tabela = arquivo.split('/')[-1].replace('.csv', '')
    
    df.to_sql(nome_tabela, engine, if_exists='replace', index=False)
    print(f"Importado: {nome_tabela} ({len(df)} linhas)")
```
