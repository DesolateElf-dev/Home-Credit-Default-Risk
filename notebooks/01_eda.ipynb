```python
DIR = 'c:/Users/emaci/VSCode/Home-Credit-Default-Risk/data/home-credit-default-risk/application_train.csv'
import pandas as pd

# Load the dataset
df = pd.read_csv(DIR)

# Verify column names and data source
print("Available columns:", df.columns.tolist())
print("First 2 rows of data:")
df.head(2)

# Display first 5 rows
df.head()

# Show statistical summary
df.describe()

# Display dataset information
df.info()
```

```python
# Install missing visualization library

import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(12, 4))

# Histogram for numerical feature
plt.subplot(1, 3, 1)
sns.histplot(df['CreditScore'], bins=30, kde=True)
plt.title('Credit Score Distribution')

# Count plot for categorical variable
plt.subplot(1, 3, 2)
sns.countplot(x='LoanStatus', data=df)
plt.title('Loan Status Counts')

# Correlation matrix
plt.subplot(1, 3, 3)
sns.heatmap(df.corr(), annot=True, fmt='.2f', cmap='coolwarm')
plt.title('Feature Correlations')

plt.tight_layout()
plt.show()
```
